```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
print(df.shape)
print(df.head())
print(df['Attrition'].value_counts())
```

```python
# How many employees left vs stayed
print(df['Attrition'].value_counts())
df['Attrition'].value_counts().plot(kind='bar', color=['steelblue','coral'], figsize=(6,4))
plt.title('Employee Attrition Count')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```

```python
dept_attrition = df[df['Attrition']=='Yes']['Department'].value_counts()

dept_attrition.plot(kind='bar', color='tomato', figsize=(8,5))
plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees Left')
plt.tight_layout()
plt.show()
```

```python
df[df['Attrition']=='Yes']['Age'].plot(kind='hist', bins=20, color='purple', figsize=(8,5))
plt.title('Age Distribution of Employees Who Left')
plt.xlabel('Age')
plt.tight_layout()
plt.show()
```

```python
avg_income = df.groupby('Attrition')['MonthlyIncome'].mean()
print(avg_income)

avg_income.plot(kind='bar', color=['green','red'], figsize=(6,4))
plt.title('Average Monthly Income: Left vs Stayed')
plt.ylabel('Monthly Income')
plt.tight_layout()
plt.show()
```

```python
wlb = df.groupby(['WorkLifeBalance','Attrition']).size().unstack()

wlb.plot(kind='bar', figsize=(8,5))
plt.title('Work Life Balance vs Attrition')
plt.xlabel('Work Life Balance (1=Bad, 4=Best)')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```
